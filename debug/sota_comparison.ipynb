```python
import matplotlib.pyplot as plt
import numpy as np

# ---------------- Daten ----------------
methods = ["Scope", "AttFusion", "S-AdaFusion", "BM2CP-RCA", "AttFuse w/ MDD", "Where2Comm-LRA"]
ap_05   = [19.6, 24.0, 38.3, 55.3,55.57, 75.1]
ap_07   = [14.4, 16.0, 30.3, 40.1,47.04, 63.1]

x = np.arange(len(methods))
width = 0.35  # Balkenbreite

# ---------------- Plot-Stil wie im Original ----------------
plt.rcParams.update(plt.rcParamsDefault)      # Standard-Matplotlib-Style
fig, ax = plt.subplots(figsize=(6, 3.6))      # ähnlich wie dein Bild

# Farben nah am Original
orange = "#FFA500"
blue   = "#4169E1"

bars1 = ax.bar(x - width/2, ap_05, width, label="AP@0.5", color=orange)
bars2 = ax.bar(x + width/2, ap_07, width, label="AP@0.7", color=blue)

# Achsen, Ticks, Labels
ax.set_ylabel("Average Precision (AP, %)", fontsize=9)
ax.set_xticks(x)
ax.set_xticklabels(methods, fontsize=8)
ax.set_ylim(0, 80)

# Legende oben links wie im Screenshot
ax.legend(loc="upper left", fontsize=8)

# Werte über den Balken schreiben
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f"{height:.1f}",
                    xy=(rect.get_x() + rect.get_width()/2, height),
                    xytext=(0, 2),            # kleiner vertikaler Offset
                    textcoords="offset points",
                    ha="center", va="bottom")

autolabel(bars1)
autolabel(bars2)

fig.tight_layout()

# ---------------- Als PDF in hoher Qualität speichern ----------------
fig.savefig("ap_barplot_cvpr.pdf",
            format="pdf",
            bbox_inches="tight",
            dpi=600)   # dpi vor allem für eingebettete Rastergrafiken

plt.close(fig)
```
